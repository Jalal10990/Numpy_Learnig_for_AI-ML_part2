```python
# The Mental Model of an Array
# A NumPy array is an organized, typed, multi-dimensional numerical grid
# Every NumPy array has 4 Sacred Attributes that describe its structure:
# Attribute    │ Meaning                                                │
# ├──────────────┼────────────────────────────────────────────────────────┤
# │ 1. ndim      │ Number of dimensions (axes). How many directions?      │
# │ 2. shape     │ A tuple showing the length along each dimension.       │
# │ 3. size      │ Total count of all numbers in the array.               │
# │ 4. dtype     │ Data type stored in memory (e.g., int32, float64). 
# --------------------------------------------------------------
# Dimensions Deep Dive: From 0D Scalar to 4D Tensor
# //0D Array (Scalar)
# Concept: Sirf aik akela single number. Koi axis/direction nahi
import numpy as np
scalar = np.array(23)

print("Array:",scalar)
print("ndim: ", scalar.ndim)
print("shape: ",scalar.shape)
print("size: ", scalar.size)
```

```python
 # 1D Array (Vector) It has only one direction (one axis).
# Concept: Numbers ki ek seedhi line (Row vector ya Feature vector).

vector = np.array([12,23,45,43])
print("ndim :", vector.ndim)  
print("shape:", vector.shape) 
print("size :", vector.size) 
# Why (3,) and not (1, 3)?
# (3,) ka matlab hai 1-dimensional array jisme 3 elements hain. Iski koi "rows ya columns" nahi hotay, yeh sirf ek 1D sequence hai.
# (1, 3) ek 2D matrix hota hai jisme 1 row aur 3 columns hotay hain. Dono memory aur math mein farq hain!
```

```python
# data types and automatic type coercion
# Because NumPy arrays are strictly homogeneous, all elements must share the exact same type. If you pass mixed data types into np.array(), NumPy will silently force (coerce) all values to the most flexible type:

mixed_arr = np.array([1,4.5,"deep_learning"])
print("Data Type:", mixed_arr.dtype)
print("Array : ", mixed_arr)
# NumPy could not convert "deep_learning" into a number, so it converted the integer 1 and float 4.5 into text strings (<U32 = 32-character Unicode string)! You can no longer do any math on this array.
```

```python
# Suppose you want to create numbers from 0 to 10.
# Question A: I want numbers that increase by 2 each time. How many numbers will I get?
# Question B: I want exactly 5 evenly spaced numbers between 0 and 10. What should those numbers be?

# np.arange()

# You choose the step size.

# 0 → 2 → 4 → 6 → 8

# The stop value 10 is excluded.
# np.linspace()

# You choose the total number of points.

# 0 → 2.5 → 5 → 7.5 → 10

# By default, the stop value 10 is included.


# arange means array range . it generates a sequence of values using a specified increments. or step size.
# start: where the sequence begins.

# stop: the boundary where the sequence stops, normally excluded.

# step: how much to increase or decrease between consecutive values.
# If we don't specify start or step, NumPy defaults to starting at zero and increasing by one.
```

```python
# linspace means linear spacing.
# it generates a soecied number of evenly spaced points between a start and stop value.
# start: first point.

# stop: last point by default.

# num: number of points to generate; defaults to 50.
# We requested exactly five points.
# 0, 2.5, 5, 7.5, 10
# Start = 0

# Stop = 10

# Number of points = 5
# Because both endpoints are included by default, there are four gaps between the five points.
# Spacing=
# num−1
# Stop−Start
# 	​

# =
# 5−1
# 10−0
# 	​

# =
# 4
# 10
# 	​

# =2.5
# linspace() calculates the step size for you so that it can produce the requested number of evenly spaced points.
```

```python
# A matrix is a rectangular arrangement of numbers in rows and columns.
# A special matrix called the identity matrix has ones on the main diagonal and zeros everywhere else.
# The diagonal runs from the top-left corner to the bottom-right corner
Position (0, 0) contains 1.

Position (1, 1) contains 1.

Position (2, 2) contains 1.

Every other position contains 0
# Create an identity matrix
# The number 3 means three rows and three columns.
# The identity matrix is square because its number of rows equals its number of columns.
# Multiplying a number by 1 doesn't change it.

# The identity matrix plays the same role in matrix multiplication.

# The @ operator performs matrix multiplication. Because we multiplied by the identity matrix, the original values were preserved.
# A @ I performs matrix multiplication and returns A.
# A * I multiplies corresponding elements, so the off-diagonal values become zero
```

```python
# Create an identity matrix with a different data type
# You can also create rectangular arrays using np.eye()
# This has three rows and four columns, so it is not a square identity matrix in the usual sense. The ones still follow the main diagonal
a = np.arange(1, 12, 3)

print(a)

```
